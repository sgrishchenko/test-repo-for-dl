```python
import pandas as pd
import random

pddf = pd.DataFrame(
    {"col1": [random.randint(0, 100) for _ in range(100)], "col2": [random.randint(0, 100) for _ in range(100)]})
pddf
```

```python
def sample(start, end, size):
    if size is None or size < 1 or size > end - start:
        return list(range(start, end))
    if size == 1:
        return [start + (end - start) // 2]
    d = (end - start) // (size - 1)
    return [start + i * d for i in range(0, size)]

rows = [r - 50 for r in sample(50, 100, 10)]

segment = pldf.select(pl.nth(range(0, 2))).slice(50, 100 - 50)
segment

[[col[row] for row in rows] for col in segment]
```
