```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a DataFrame with datetime values in non-chronological order
data = {
    "date": [
        "2025-02-10", "2025-02-05", "2025-02-15", "2025-02-08", "2025-02-12"
    ],
    "value": [10, 15, 7, 20, 12]
}

df = pd.DataFrame(data)
df["date"] = pd.to_datetime(df["date"])  # Convert to datetime

df
```

```python
from lets_plot import * 
ggplot() + \
geom_line(aes(x="date", y="value"), data=df, sampling="none" if df.size < 2500 else sampling_systematic(n=2500)) + \
ggtitle("Chart") 
```
