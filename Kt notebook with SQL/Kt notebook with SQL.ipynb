```kotlin
import krangl.DataFrame
import krangl.dataFrameOf

// Create a DataFrame with columns 'col1' and 'col2'
val df = dataFrameOf("col1", "col2")(
    1, 4,
    2, 6,
    3, 5
)

// Display the DataFrame
println(df)
```
